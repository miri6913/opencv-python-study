# 1. putText()

### Task 1: Write Your Name

```python
import cv2
import numpy as np

img = np.zeros((500, 500, 3), dtype = np.uint8)
sample1 = img.copy()

cv2.putText(sample1, "Mike", (100, 100), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 3)

cv2.imshow("sample1", sample1)
cv2.waitKey()
cv2.destroyAllWindows()
```

### Task 2: Change Font Size

```python
import cv2
import numpy as np

img = np.zeros((500, 500, 3), dtype = np.uint8)
sample1 = img.copy()

cv2.putText(sample1, "Mike", (0, 250), cv2.FONT_HERSHEY_SIMPLEX, 3, (255, 255, 255), 3)

cv2.imshow("sample1", sample1)
cv2.waitKey()
cv2.destroyAllWindows()
```

### Task 3: Change Color

```python
import cv2
import numpy as np

img = np.zeros((500, 500, 3), dtype = np.uint8)
sample1 = img.copy()

cv2.putText(sample1, "Mike", (0, 250), cv2.FONT_HERSHEY_SIMPLEX, 3, (0, 255, 255), 3)

cv2.imshow("sample1", sample1)
cv2.waitKey()
cv2.destroyAllWindows()
```

### Task 4: Create a Watermark

```python
sample2 = img.copy()

cv2.putText(sample2, "0 Mike 2026", (0, 250), cv2.FONT_HERSHEY_COMPLEX, 1, (255, 255, 255), 3)

cv2.imshow("sample2", sample2)
cv2.waitKey()
cv2.destroyAllWindows()
```

### Task 5: Display Image Size

```python
x, y, _ = img.shape

sample3 = img.copy()

cv2.putText(sample3, str(x), (0, 250), cv2.FONT_HERSHEY_COMPLEX, 1, (255, 255, 255), 3)
cv2.putText(sample3, str(y), (0, 300), cv2.FONT_HERSHEY_COMPLEX, 1, (255, 255, 255), 3)

cv2.imshow("sample3", sample3)
cv2.waitKey()
cv2.destroyAllWindows()
```
