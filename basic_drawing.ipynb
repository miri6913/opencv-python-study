A good way to learn OpenCV drawing functions is to solve progressively harder tasks. Try to write the code yourself first, then verify the output visually.

---

# 1. cv2.line()

### Task 1: Horizontal Line

Create a 500×500 black image and draw a white horizontal line across the center.

Goal:

* Learn start point and end point.

---

### Task 2: Cross Shape

Draw:

* One vertical line
* One horizontal line

so they intersect in the middle.

Goal:

* Practice multiple lines.

---

### Task 3: House Roof

Draw a triangle roof using 3 separate lines.

Example:

```
   /\
  /  \
 /____\
```

Goal:

* Connect multiple line segments.

---

### Task 4: Grid

Create a 5×5 grid using only lines.

Goal:

* Learn loops with line().

---

### Task 5: Clock Hands

Draw a clock with:

* Center point
* Hour hand
* Minute hand

using lines only.

Goal:

* Use geometry and coordinates.

---

# 2. cv2.ellipse()

### Task 1: Simple Ellipse

Draw one ellipse in the center.

Goal:

* Learn:

```python
cv2.ellipse(...)
```

parameters.

---

### Task 2: Vertical vs Horizontal Ellipse

Draw:

* Wide ellipse
* Tall ellipse

side by side.

Goal:

* Understand axes parameter.

---

### Task 3: Rotated Ellipse

Draw the same ellipse rotated:

* 0°
* 45°
* 90°

Goal:

* Learn rotation angle.

---

### Task 4: Smiley Mouth

Draw a smile using an ellipse arc.

Example:

```
  :)
```

Goal:

* Learn startAngle and endAngle.

---

### Task 5: Olympic-style Logo

Create 5 overlapping ellipses.

Goal:

* Practice positioning and sizing.

---

# 3. cv2.rectangle()

### Task 1: Single Rectangle

Draw one rectangle in the center.

Goal:

* Learn top-left and bottom-right points.

---

### Task 2: Frame Border

Draw a border around the entire image.

Goal:

* Understand image coordinates.

---

### Task 3: Chessboard Blocks

Draw 8 rectangles in a row.

Goal:

* Use loops.

---

### Task 4: Building Windows

Draw a building and fill it with rows of windows (small rectangles).

Goal:

* Practice nested loops.

---

### Task 5: Bounding Boxes

Given these coordinates:

```python
boxes = [
    (50,50,100,120),
    (200,100,150,80),
    (100,250,120,120)
]
```

Draw all boxes.

Goal:

* Simulate object detection output.

# Final Challenge (Uses Everything)

Create a small city scene containing:

* Sun (circle)
* Road (filled polygon)
* Buildings (rectangles)
* Windows (rectangles)
* Trees (circles + rectangles)
* Power lines (lines)
* Clouds (ellipses)

If you can build this scene from scratch without looking at notes, you've probably memorized all five functions.
