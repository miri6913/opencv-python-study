## 1. cv2.line()
### Task 1: Horizontal Line
Create a 500×500 black image and draw a white horizontal line across the center.

````python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = np.zeros((500, 500, 3))
img1 = img.copy()
cv2.line(img1,(0, 250), (500,250), (0, 255, 255), 5)

cv2.imshow("img1", img1)
cv2.waitKey()
cv2.destroyAllWindows()

#plt.imshow(img1)
#plt.title('smth')
````

### Task 2: Cross Shape
Draw:
* One vertical line
* One horizontal line
so they intersect in the middle.


````python
img = np.zeros((500, 500, 3), dtype=np.uint8)
img2 = img.copy()
cv2.line(img2, (0, 250), (500, 250), (0, 0, 255), 5)
cv2.line(img2, (250, 0), (250, 500), (0, 255, 0), 5)

cv2.imshow("x", img2)
cv2.waitKey()
cv2.destroyAllWindows()
````

### Task 3: House Roof
Draw a triangle roof using 3 separate lines.

````python
img3 = img.copy()
pt1 = (250, 100)
pt2 = (150, 300)

pt3 = (350, 300)


color = (255, 255, 255)

cv2.line(img3, pt1, pt2, color, 3)
cv2.line(img3, pt1, pt3, color, 3)
cv2.line(img3, pt2, pt3, color, 3)

cv2.imshow("img3", img3)
cv2.waitKey()
cv2.destroyAllWindows()
````

### Task 4: Grid
Create a 5×5 grid using only lines.

````python
img5 = img.copy()

for i in range(6):
   tmp = i*100
   cv2.line(img5, (0, 50+tmp), (500, 50+tmp), color, 3) 
   
for i in range(6):
   tmp = i*100
   cv2.line(img5, (50+tmp, 0), (50+tmp, 500), color, 3)
   
cv2.imshow("img5", img5)
cv2.waitKey()
cv2.destroyAllWindows()
````

### Task 5: Clock Hands
Draw a clock with:
* Center point
* Hour hand
* Minute hand
using lines only.

````python
img6 = img.copy()

cv2.line(img6, (250, 250), (250, 250), color, 7)
cv2.line(img6, (250, 250), (400, 40), color, 3)
cv2.line(img6, (250, 250), (400, 150), color, 3)

cv2.imshow("img6", img6)
cv2.waitKey()
cv2.destroyAllWindows()
````

### Task 5: Bounding Boxes
Given these coordinates:
```python
boxes = [
    (50,50,100,120),
    (200,100,150,80),
    (100,250,120,120)
]
```
Draw all boxes.

````python
rec5 = img.copy()

boxes = [
    (50,50,100,120),
    (200,100,150,80),
    (100,250,120,120)
]

for box in boxes:
    x1, y1, x2, y2 = box
    cv2.rectangle(rec5, (x1, y1), (x1+x2, x2+y2), color, 3)
    
cv2.imshow("rec5", rec5)
cv2.waitKey()
cv2.destroyAllWindows()
````

### Task 5: Bubble Generator
Generate 100 random circles.

````python
import numpy as np

img_c3 = img.copy()

for i in range(100):
    radius = np.random.randint(3, 50)
    
    x = np.random.randint(radius, img_c3.shape[1] - radius)
    y= np.random.randint(radius, img_c3.shape[0] - radius)
    
    color = (
        np.random.randint(0, 256),
        np.random.randint(0, 256),
        np.random.randint(0, 256),
    )
    print(color)
    cv2.circle(img_c3, (x, y), radius, color, 3)
    
cv2.imshow("img_c3", img_c3)
cv2.waitKey()
cv2.destroyAllWindows()
````
