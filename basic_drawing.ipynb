````python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = np.zeros((500, 500, 3))
cv2.line(img,(0, 250), (500,250), (0, 255, 255), 5)

cv2.imshow("img", img)
cv2.waitKey()
cv2.destroyAllWindows()

#plt.imshow(img)
#plt.title('smth')
````

````python
img = np.zeros((500, 500, 3), dtype=np.uint8)
img2 = img.copy()
cv2.line(img2, (0, 250), (500, 250), (0, 0, 255), 5)
cv2.line(img2, (250, 0), (250, 500), (0, 255, 0), 5)

cv2.imshow("x", img2)
cv2.waitKey()
cv2.destroyAllWindows()
````

````python
img3 = img.copy()
pt1 = (250, 100)
pt2 = (150, 300)

pt3 = (350, 300)


color = (255, 255, 255)

cv2.line(img3, pt1, pt2, color, 3)
cv2.line(img3, pt1, pt3, color, 3)
cv2.line(img3, pt2, pt3, color, 3)

cv2.imshow("img3", img3)
cv2.waitKey()
cv2.destroyAllWindows()
````

````python
img5 = img.copy()

for i in range(6):
   tmp = i*100
   cv2.line(img5, (0, 50+tmp), (500, 50+tmp), color, 3) 
   
for i in range(6):
   tmp = i*100
   cv2.line(img5, (50+tmp, 0), (50+tmp, 500), color, 3)
   
cv2.imshow("img5", img5)
cv2.waitKey()
cv2.destroyAllWindows()
````
